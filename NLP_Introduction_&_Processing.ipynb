## Why Can a Machine Not Directly Understand Raw Text?

A computer fundamentally processes numerical representations, while raw human language consists of words, sentences, symbols, and linguistic structures.

For example:

```text
"I love Python."
```

A computer cannot directly perform mathematical operations on the meaning of this sentence without converting the text into a suitable numerical representation.

Therefore, NLP converts text into structured or numerical representations that machine-learning algorithms can process.

# What are Word Embeddings?

**Word embeddings** are dense numerical vector representations of words.

Words with similar meanings are generally represented by vectors that are closer together in the embedding space.

For example:

```text
king
queen
man
woman
```

can be represented as dense vectors.

---

## Word Embeddings vs Bag of Words vs TF-IDF

| Feature                | Bag of Words    | TF-IDF               | Word Embeddings                                             |
| ---------------------- | --------------- | -------------------- | ----------------------------------------------------------- |
| Representation         | Sparse          | Sparse               | Dense                                                       |
| Semantic relationships | Very limited    | Limited              | Better                                                      |
| Vector size            | Vocabulary size | Vocabulary size      | Fixed embedding dimension                                   |
| Word order             | Usually ignored | Usually ignored      | Basic Word2Vec is context-based but not full sentence order |
| Meaning similarity     | Poor            | Limited              | Better                                                      |
| Example                | Word counts     | Weighted word counts | Dense vectors                                               |

### Bag of Words

Represents a document using word occurrence/counts.

Example:

```text
"I love NLP"
```

may become a vector based on vocabulary counts.

### TF-IDF

TF-IDF gives higher importance to words that are frequent in a document but relatively uncommon across the document collection.

### Word Embeddings

Word embeddings represent words using dense numerical vectors that can capture useful semantic and contextual relationships.

### Easy Definition:

**BoW and TF-IDF mainly represent word importance in documents, while word embeddings represent words as dense vectors that can capture semantic relationships.**

## What is Sequential Data? Explain why an ordinary feed-forward neural network is not sufficient for many NLP tasks.

### Answer:

**Sequential data** is data in which the order of observations or elements is important.

Examples include:

* Sentences
* Text documents
* Speech
* Time series
* Stock prices
* Sensor readings

For example:

```text
"I love Python"
```

has a different structure from:

```text
"Python love I"
```

The order of words contributes to meaning.

---

## Why Feed-Forward Neural Networks Can Be Insufficient

A traditional feed-forward neural network generally processes its input without maintaining a hidden state representing previous sequence elements.

For many NLP tasks, understanding the current word requires information from previous or later words.

### Example:

```text
"The bank is near the river."
```

versus:

```text
"The bank approved my loan."
```

The word **bank** has different meanings depending on context.

A feed-forward network does not naturally maintain sequential context.

Therefore, models such as:

* RNN
* LSTM
* GRU
* Transformers

are commonly used for sequence-related tasks.

---

# Compare RNN, LSTM and GRU

## RNN – Recurrent Neural Network

RNN processes sequential data while maintaining a hidden state.

### Advantages:

* Simple architecture.
* Can process sequences.
* Maintains information from previous time steps.

### Limitation:

Standard RNNs can suffer from **vanishing and exploding gradient problems**, making long-term dependencies difficult to learn.

---

## LSTM – Long Short-Term Memory

LSTM is a specialized RNN architecture designed to better handle long-term dependencies.

It uses gates such as:

* Forget gate
* Input gate
* Output gate

### Advantages:

* Handles long-term dependencies better than basic RNN.
* Reduces the impact of vanishing gradients.

### Limitation:

* More parameters.
* More computationally expensive than simple RNN.

---

## GRU – Gated Recurrent Unit

GRU is another gated recurrent architecture.

It uses two main gates:

* Update gate
* Reset gate

### Advantages:

* Simpler than LSTM.
* Fewer parameters.
* Often faster to train than LSTM.

### Comparison:

| Feature                       | RNN              | LSTM             | GRU                     |
| ----------------------------- | ---------------- | ---------------- | ----------------------- |
| Basic recurrent architecture  | Yes              | Yes              | Yes                     |
| Gates                         | No               | 3 main gates     | 2 main gates            |
| Long-term dependency handling | Weak             | Stronger         | Strong                  |
| Parameters                    | Fewer            | More             | Usually fewer than LSTM |
| Complexity                    | Low              | High             | Medium                  |
| Training speed                | Generally faster | Generally slower | Often faster than LSTM  |

# Evaluation Metrics

## 1. Accuracy

Accuracy measures the proportion of total predictions that are correct.

$$
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
$$

---

## 2. Precision

Precision measures how many predicted positive cases are actually positive.

$$
Precision =
\frac{TP}
{TP+FP}
$$

---

## 3. Recall

Recall measures how many actual positive cases were correctly identified.

$$
Recall =
\frac{TP}
{TP+FN}
$$

---

## 4. F1-Score

F1-score is the harmonic mean of precision and recall.

$$
F1 =
2\times
\frac{Precision\times Recall}
{Precision+Recall}
$$

---

## 5. Perplexity

**Perplexity** is commonly used to evaluate language models.

Lower perplexity generally indicates that the model assigns higher probability to the observed sequence.

Conceptually:

$$
Perplexity =
e^{-\frac{1}{N}
\sum_{i=1}^{N}
\log P(w_i)}
$$

A lower value generally indicates better predictive performance for the evaluated language-modeling task.

### Easy Definition:

**RNN, LSTM and GRU process sequential information; LSTM and GRU use gates to handle long-term dependencies better than basic RNNs.**

# QUESTION 5

## What is a Sequence-to-Sequence model? Explain the roles of the encoder, decoder, hidden states, and context vector.

### Answer:

A **Sequence-to-Sequence (Seq2Seq) model** is a neural-network architecture that converts one sequence into another sequence.

It is commonly used for:

* Machine translation
* Text summarization
* Question answering
* Conversational systems

### Example:

```text
English Sentence
       ↓
    Encoder
       ↓
Context / Representation
       ↓
    Decoder
       ↓
Hindi Sentence
```

---

## 1. Encoder

The encoder processes the input sequence one element at a time and converts it into an internal representation.

Example:

```text
"I love Python"
```

The encoder processes:

```text
I → love → Python
```

---

## 2. Decoder

The decoder generates the output sequence step by step.

For example:

```text
I love Python
```

may be translated into another language by generating output tokens sequentially.

---

## 3. Hidden States

Hidden states contain information learned from previous sequence elements.

They allow the recurrent network to maintain information while processing a sequence.

---

## 4. Context Vector

In a basic encoder-decoder model, the final encoder representation can act as a **context vector** summarizing the input sequence.

The decoder uses this representation to generate the output.

---

# Why Does Basic Seq2Seq Struggle with Long Sentences?

A basic encoder-decoder architecture may need to compress the entire input sequence into a fixed-size representation.

For a long sentence, important information from the beginning may be difficult to preserve.

This creates an information bottleneck.

---

# Attention Mechanism

The **attention mechanism** solves this problem by allowing the decoder to access different encoder hidden states when generating each output token.

Instead of relying only on one fixed context vector, the decoder can assign different weights to different input positions.

### Example:

```text
Input:
"The cat is sitting on the mat."

             ↓

Decoder step 1
Focus → "The"

Decoder step 2
Focus → "cat"

Decoder step 3
Focus → "sitting"

...
```

The model dynamically focuses on relevant parts of the input.

### Basic Attention Process:

```text
Encoder Hidden States
      ↓
Attention Scores
      ↓
Softmax
      ↓
Attention Weights
      ↓
Weighted Sum
      ↓
Context Vector
      ↓
Decoder
```

### Advantages of Attention:

* Handles long sequences better.
* Reduces the fixed-vector bottleneck.
* Allows the model to focus on relevant input words.
* Improves sequence-to-sequence performance.

### Easy Definition:

**Seq2Seq converts one sequence into another, while attention allows the decoder to focus on the most relevant input positions at each step.**

## Building a Reusable Text-Preprocessing Function

### Given Text:

```text
" This PRODUCT is Amazing!!! I bought 2 items, and I really loved them. "
```

The program should:

* Convert text to lowercase.
* Remove punctuation.
* Remove numbers.
* Remove extra whitespaces.
* Remove English stopwords.
* Tokenize the cleaned text.
* Create a reusable preprocessing function.

### Google Colab Code:

```python
# Install NLTK if required
!pip install -q nltk
```

```python
import nltk
import re
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

def preprocess_text(text):

    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = text.translate(
        str.maketrans(
            "",
            "",
            string.punctuation
        )
    )

    # 3. Remove numbers
    text = re.sub(
        r"\d+",
        "",
        text
    )

    # 4. Remove extra whitespaces
    text = " ".join(
        text.split()
    )

    # 5. Tokenization
    tokens = word_tokenize(text)

    # 6. Remove stopwords
    stop_words = set(
        stopwords.words("english")
    )

    tokens = [
        word
        for word in tokens
        if word not in stop_words
    ]

    return tokens


# Given text
text = (
    " This PRODUCT is Amazing!!! "
    "I bought 2 items, and I really loved them. "
)

# Apply function
cleaned_tokens = preprocess_text(text)

print("Cleaned Tokens:")
print(cleaned_tokens)
```

### Expected Output:

The output will contain cleaned tokens similar to:

```text
['product', 'amazing', 'bought', 'items', 'really', 'loved']
```

The exact stopword result depends on the NLTK English stopword list/version.

### Explanation:

The function is reusable because the same `preprocess_text()` function can be applied to multiple text documents.

### Easy Definition:

**A reusable preprocessing function performs the same text-cleaning steps consistently on different inputs.**

# QUESTION 7

## Use NLTK to tokenize the sentence, apply Porter stemming, WordNet lemmatization, and generate unigrams, bigrams, and trigrams.

### Given Sentence:

```text
"The students were studying and running towards their classes."
```

### Google Colab Code:

```python
!pip install -q nltk
```

```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.util import ngrams
from nltk import pos_tag

# Download resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")

# Sentence
sentence = (
    "The students were studying and running "
    "towards their classes."
)

# Tokenization
tokens = word_tokenize(
    sentence
)

print("Tokens:")
print(tokens)

# Porter Stemming
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in tokens
]

print("\nPorter Stemming:")
print(stemmed_words)

# POS tagging
pos_tags = pos_tag(tokens)

print("\nPOS Tags:")
print(pos_tags)

# Convert POS tags to WordNet POS
from nltk.corpus import wordnet

def get_wordnet_pos(tag):

    if tag.startswith("J"):
        return wordnet.ADJ

    elif tag.startswith("V"):
        return wordnet.VERB

    elif tag.startswith("N"):
        return wordnet.NOUN

    elif tag.startswith("R"):
        return wordnet.ADV

    return wordnet.NOUN


# WordNet Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(
        word,
        get_wordnet_pos(tag)
    )
    for word, tag in pos_tags
]

print("\nWordNet Lemmatization:")
print(lemmatized_words)

# Unigrams
unigrams = list(
    ngrams(tokens, 1)
)

print("\nUnigrams:")
print(unigrams)

# Bigrams
bigrams = list(
    ngrams(tokens, 2)
)

print("\nBigrams:")
print(bigrams)

# Trigrams
trigrams = list(
    ngrams(tokens, 3)
)

print("\nTrigrams:")
print(trigrams)
```

---

## Comparison of Stemming and Lemmatization

### Stemming

Stemming uses rule-based word reduction.

For example:

```text
studying → studi
running → run
classes → class
```

A stem may not always be a valid dictionary word.

### Lemmatization

Lemmatization attempts to produce a meaningful dictionary base form.

Examples can include:

```text
students → student
studying → study
running → run
classes → class
```

The exact result depends on the POS information supplied to the lemmatizer.

### Main Difference:

| Stemming                    | Lemmatization                          |
| --------------------------- | -------------------------------------- |
| Faster                      | Usually more computationally involved  |
| Rule-based reduction        | Uses linguistic/dictionary information |
| May produce non-words       | Usually produces valid base forms      |
| Less linguistically precise | More linguistically meaningful         |

### Easy Definition:

**Stemming cuts words to a root-like form, while lemmatization tries to find the correct dictionary base form.**

## Use the following documents to create Bag of Words and TF-IDF representations.

### Documents:

```text
Document 1:
"NLP is useful for text analysis"

Document 2:
"Text analysis is useful for business"

Document 3:
"NLP converts text into numerical features"
```

---

## Part A: Bag of Words

### Python Code:

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "NLP is useful for text analysis",
    "Text analysis is useful for business",
    "NLP converts text into numerical features"
]

# Create Bag of Words
bow_vectorizer = CountVectorizer()

X_bow = bow_vectorizer.fit_transform(
    documents
)

# Vocabulary
print("Vocabulary:")
print(
    bow_vectorizer.vocabulary_
)

# Feature names
print("\nFeature Names:")
print(
    bow_vectorizer.get_feature_names_out()
)

# Feature matrix
print("\nBag of Words Matrix:")
print(
    X_bow.toarray()
)
```

### Explanation:

Bag of Words represents each document as a vector containing word counts.

---

# Part B: TF-IDF

### Python Code:

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF
tfidf_vectorizer = TfidfVectorizer()

X_tfidf = tfidf_vectorizer.fit_transform(
    documents
)

# Vocabulary
print("TF-IDF Vocabulary:")
print(
    tfidf_vectorizer.get_feature_names_out()
)

# TF-IDF matrix
print("\nTF-IDF Matrix:")

print(
    X_tfidf.toarray()
)
```

---

# Identify the Word with Highest TF-IDF Score in Each Document

```python
import numpy as np

feature_names = (
    tfidf_vectorizer
    .get_feature_names_out()
)

tfidf_matrix = X_tfidf.toarray()

for i, row in enumerate(
    tfidf_matrix
):

    max_index = np.argmax(row)

    word = feature_names[
        max_index
    ]

    score = row[
        max_index
    ]

    print(
        f"Document {i+1}: "
        f"{word} "
        f"(TF-IDF = {score:.4f})"
    )
```

### Important Note:

The exact highest-scoring word should be taken from the program output because TF-IDF values depend on the vectorizer settings and corpus.

### Interpretation:

* **TF** measures how frequently a word occurs in a document.
* **IDF** reduces the importance of words that occur across many documents.
* **TF-IDF** gives higher weight to words that are important to a document relative to the collection.

### Formula:

$$
TFIDF(t,d)=TF(t,d)\times IDF(t)
$$

A common IDF form is:

$$
IDF(t)=\log
\left(
\frac{N+1}{df(t)+1}
\right)+1
$$

where:

* \(N\) = number of documents
* \(df(t)\) = number of documents containing term \(t\)

### Easy Definition:

**Bag of Words counts words, while TF-IDF gives greater importance to words that are relatively distinctive within a document collection.**

### Step 2: Python Code

```python
from gensim.models import Word2Vec

# Small training corpus
sentences = [
    ["king", "queen", "royal", "palace"],
    ["king", "man", "prince", "royal"],
    ["queen", "woman", "princess", "royal"],
    ["man", "woman", "person", "human"],
    ["prince", "king", "royal", "male"],
    ["princess", "queen", "royal", "female"],
    ["king", "queen", "prince", "princess"],
    ["royal", "palace", "king", "queen"],
    ["man", "boy", "king", "male"],
    ["woman", "girl", "queen", "female"]
]

# Train Word2Vec
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    epochs=200,
    seed=42
)

# Similar words
print(
    "Words similar to 'king':"
)

print(
    model.wv.most_similar(
        "king",
        topn=5
    )
)
```

### Explanation:

Word2Vec learns dense vector representations based on word context.

The `most_similar()` method finds words whose vectors are close to the vector of `"king"`.

### Important Note:

Because this is a **very small artificial training corpus**, the exact similarity values and ranking are not expected to represent real-world language relationships. A large corpus would produce more meaningful embeddings.

### Easy Definition:

**Word2Vec learns word vectors from surrounding context, allowing similar words to have similar vector representations.**

# QUESTION 10

# PART A: LSTM Model

## Build a simple LSTM model in TensorFlow/Keras for binary sentiment classification.

The model should contain:

* Text vectorization/tokenization
* Embedding layer
* LSTM layer
* Dense output layer

### Answer:

We can use the IMDB movie-review dataset available through Keras.

The target is binary:

```text
0 → Negative
1 → Positive
```

### Google Colab Code:

```python
import tensorflow as tf

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.datasets import imdb
```

---

## Load Dataset

```python
vocab_size = 10000

max_length = 200

(X_train, y_train), (
    X_test,
    y_test
) = imdb.load_data(
    num_words=vocab_size
)
```

---

## Pad Sequences

The reviews have different lengths, so we convert them to a fixed length.

```python
X_train = tf.keras.utils.pad_sequences(
    X_train,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

X_test = tf.keras.utils.pad_sequences(
    X_test,
    maxlen=max_length,
    padding="post",
    truncating="post"
)
```

---

## Build LSTM Model

```python
model = models.Sequential([

    layers.Embedding(
        input_dim=vocab_size,
        output_dim=64,
        input_length=max_length
    ),

    layers.LSTM(64),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])
```

---

## Compile Model

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

---

## Display Model Architecture

```python
model.summary()
```

---

## Train Model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=3,
    batch_size=64,
    validation_split=0.2
)
```

---

## Evaluate Model

```python
test_loss, test_accuracy = (
    model.evaluate(
        X_test,
        y_test
    )
)

print(
    "Test Accuracy:",
    test_accuracy
)
```

### Model Architecture:

```text
Text / Tokenized Review
          ↓
Embedding Layer
          ↓
LSTM Layer
          ↓
Dense Layer
          ↓
Sigmoid
          ↓
Positive / Negative
```

### Explanation of Layers

#### 1. Embedding Layer

Converts integer word IDs into dense vectors.

#### 2. LSTM Layer

Processes the sequence and learns information from previous tokens.

#### 3. Dense Layer

Produces the final binary classification output.

#### 4. Sigmoid Activation

Produces a value between 0 and 1 that can be interpreted as the estimated probability of the positive class.

### Easy Definition:

**An LSTM sentiment model converts words into embeddings, processes their sequence using LSTM, and predicts positive or negative sentiment using a Dense output layer.**

# PART B: ATTENTION CALCULATION

## Given Attention Scores:

```text
[1.2, 0.5, 2.0]
```

We need to calculate the attention weights using **Softmax**.

### Softmax Formula:

$$
Softmax(x_i)
=
\frac{e^{x_i}}
{\sum_j e^{x_j}}
$$

Given:

$$
x=[1.2,0.5,2.0]
$$

### Step 1: Calculate Exponentials

$$
e^{1.2}\approx3.3201
$$

$$
e^{0.5}\approx1.6487
$$

$$
e^{2.0}\approx7.3891
$$

### Step 2: Calculate Sum

$$
3.3201+1.6487+7.3891
\approx12.3579
$$

### Step 3: Calculate Attention Weights

For the first position:

$$
\frac{3.3201}{12.3579}
\approx0.2687
$$

For the second position:

$$
\frac{1.6487}{12.3579}
\approx0.1334
$$

For the third position:

$$
\frac{7.3891}{12.3579}
\approx0.5979
$$

### Final Attention Weights:

```text
[0.2687, 0.1334, 0.5979]
```

### Sum:

$$
0.2687+0.1334+0.5979
\approx1
$$

Therefore:

```text
Position 1 → 0.2687
Position 2 → 0.1334
Position 3 → 0.5979
```

### Highest Attention:

The **third input position** receives the highest attention because:

$$
0.5979 > 0.2687 > 0.1334
$$

Therefore:

**Highest attention = Position 3**

### Python Verification:

```python
import numpy as np

scores = np.array([
    1.2,
    0.5,
    2.0
])

exp_scores = np.exp(
    scores
)

attention_weights = (
    exp_scores /
    np.sum(exp_scores)
)

print(
    "Attention Weights:",
    attention_weights
)

print(
    "Highest Attention Position:",
    np.argmax(
        attention_weights
    ) + 1
)
```

### Expected Output:

```text
Attention Weights:
[0.26866364 0.13341442 0.59792194]

Highest Attention Position:
3
```
